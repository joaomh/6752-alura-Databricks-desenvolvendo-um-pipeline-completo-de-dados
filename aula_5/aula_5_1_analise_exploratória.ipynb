## Volumetria geral

```python
%sql
SELECT 
    COUNT(id_interacao) AS total_atendimentos,
    COUNT(DISTINCT cliente_id) AS clientes_unicos,
    MIN(data_hora) AS primeira_interacao,
    MAX(data_hora) AS ultima_interacao
FROM prod.gold.atendimentos_alumax;
```

## Distribuição e Preferência de Canais e Departamentos

```python
%sql
SELECT 
    canal,
    departamento,
    COUNT(id_interacao) AS qtd_atendimentos,
    ROUND(COUNT(id_interacao) * 100.0 / SUM(COUNT(id_interacao)) OVER(), 2) AS percentual_total
FROM prod.gold.atendimentos_alumax
GROUP BY canal, departamento
ORDER BY qtd_atendimentos DESC;
```

## Qualidade: Avaliação por Canal e Departamento

```python
%sql
SELECT 
    canal,
    departamento,
    ROUND(AVG(satisfacao_pre_atendimento), 2) AS media_satisfacao,
    COUNT(id_interacao) AS volume_avaliacoes
FROM prod.gold.atendimentos_alumax
WHERE satisfacao_pre_atendimento IS NOT NULL
GROUP BY canal, departamento
ORDER BY media_satisfacao ASC; 
```
